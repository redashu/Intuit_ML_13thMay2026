```python
# load data
df=pd.read_csv('https://raw.githubusercontent.com/redashu/Intuit_ML_13thMay2026/refs/heads/master/datasets/customer_churn_1000.csv')
```

```python
df.info()
```

```python
print("=== Print Missing values ")
df.isnull().sum()
```

```python
# checking statistical summary
df.describe()
```

```python
# using seaborn do data analysis and EDA options
sb.countplot(x='Churn',data=df)
plt.title("Churn Distribution")
plt.show()
```

```python
# doing feature corelation of churn vs tenure
sb.boxplot(x='Churn',y='tenure',data=df)
plt.title("Tenure vs Churn")
plt.show()
```

```python
# doing feature corelation of churn vs tenure
sb.boxplot(x='Churn',y='MonthlyCharges',data=df)
plt.title("MOnthlycharges vs Churn")
plt.show()
```

```python
# using seaborn do data analysis and EDA options
sb.countplot(x='Contract',
             hue='Churn',
             data=df)
plt.title("contract type vs  churn")
plt.show()
```

```python
sb.countplot(x='InternetService',
             hue='Churn',
             data=df)
plt.title("Internetsvc type vs  churn")
plt.show()
```

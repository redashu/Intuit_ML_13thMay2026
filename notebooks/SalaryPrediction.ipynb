```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load data
df=pd.read_csv('https://raw.githubusercontent.com/redashu/Datasets/refs/heads/master/employee_salary_regression_dataset.csv')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# statiscally analysis
df.describe()
```

```python
# salary distribution using histplot
plt.title('Salary Distribution')
sns.histplot(df['Salary'],kde=True)
plt.show()
```

```python
## Detect outlier using boxplot
plt.title('Salary Distribution')
sns.boxplot(df['Salary'])
plt.show()
```

```python
# plotting graph experience vs salary
plt.title('Experience vs Salary')
sns.scatterplot(x='Experience',
                y='Salary',
                data=df)
plt.show()
```

```python
# Education vs salary using boxplot
plt.title('Education vs Salary')
sns.boxplot(x='Education',
            y='Salary',
            data=df)
plt.show()
```

```python
# Department vs salary using boxplot
plt.title('Department vs Salary')
sns.boxplot(x='Department',
            y='Salary',
            data=df)
plt.show()
```
